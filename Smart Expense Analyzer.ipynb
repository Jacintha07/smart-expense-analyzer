# Smart Expense Analyzer
A Python Project that analyzes monthly expenses using lists, tuples and loops.

```python
Expenses=[]                            #Data input
n=int(input("How many Expenses:"))
for i in range(n):
    Name=input("Name:")
    Category=input("Category:")
    Amount=int(input("Amount:"))
    Expenses.append((Name,Category,Amount))
print(Expenses)
total=0                                 #Total Expenses
for item in Expenses:
    total=total+item[2]
print("Total Expense:",total)
food_total=0                            #Category wise total
travel_total=0
shopping_total=0
for item in Expenses:
    if item[1]=="Food":
        food_total+=item[2]
    elif item[1]=="Travel":
        travel_total+=item[2]
    elif item[1]=="Shopping":
        shopping_total+=item[2]
print("Food Total",food_total)
print("Travel Total",travel_total)
print("Shopping Total",shopping_total)        
highest=Expenses[0]                     #Highest Expense
for item in Expenses:
    if item[2]>highest[2]:
        highest=item
print("Highest Expenses:",highest[0],"-",highest[2])
max_category="Food"                     #Insight
max_amount=food_total
if travel_total>max_amount:             #720>280 true
    max_category="Travel" 
    max_amount=travel_total
if shopping_total>max_amount:           #800>720 true
    max_category="Shopping"
    max_amount=shopping_total
percent=(max_amount/total)*100
print(f"Insight: You spend the most on {max_category} - {round(percent)} % of your total")
```
